```python
import torch
from torch.utils.data import Dataset, DataLoader

# Class labels
CLASSES = ["Regular", "Limping", "Shuffling"]

CLASS_TO_IDX = {
    "Regular": 0,
    "Limping": 1,
    "Shuffling": 2
}

class GaitTensorDataset(Dataset):

    def __init__(self, files):
        self.files = files

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        path = self.files[idx]

        # Determine label from the parent folder
        class_name = path.parent.name
        label = CLASS_TO_IDX[class_name]

        # Load the preprocessed tensor
        frames = torch.load(
            path,
            map_location="cpu",
            weights_only=True
        )

        # Verify tensor shape
        if frames.shape != (3, 32, 256, 256):
            raise ValueError(
                f"Unexpected shape in {path}: {frames.shape}"
            )

        if not torch.isfinite(frames).all():
            raise ValueError(
                f"Non-finite values found in {path}"
            )

        return frames.float(), label


# Create datasets
train_dataset = GaitTensorDataset(train_files)
val_dataset = GaitTensorDataset(val_files)
test_dataset = GaitTensorDataset(test_files)

# Batch size
BATCH_SIZE = 2

# Reproducibility
SEED = 42
generator = torch.Generator()
generator.manual_seed(SEED)

# Create DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
    generator=generator,
    drop_last=False
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
    drop_last=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
    drop_last=False
)

print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn
from pytorchvideo.models.hub import slowfast_r50

# Select GPU if available
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

if device.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# Load pretrained SlowFast-R50
model = slowfast_r50(pretrained=True)

# Freeze pretrained parameters
for param in model.parameters():
    param.requires_grad = False

# Replace the final classification layer
old_head = model.blocks[-1].proj

model.blocks[-1].proj = nn.Linear(
    old_head.in_features,
    len(CLASSES)
)

# =====================================================================
# CORRECTED FOR EXPERIMENT 2: Fine-tune Last Residual Stage + Head
# Targets: Outer Block 4 (ResNet Stage 5) and Outer Block 6 (Head Classifier)
# Bypasses: Outer Block 5 (PoolConcatPathway) because it has 0 weights.
# =====================================================================
# Unfreeze final residual stage
for param in model.blocks[4].parameters():
    param.requires_grad = True

# Unfreeze new classifier
for param in model.blocks[6].proj.parameters():
    param.requires_grad = True
# =====================================================================

# Move model to GPU
model = model.to(device)

# Count parameters
total_params = sum(
    p.numel() for p in model.parameters()
)

trainable_params = sum(
    p.numel() for p in model.parameters()
    if p.requires_grad
)

print("Total parameters:", f"{total_params:,}")
print("Trainable parameters:", f"{trainable_params:,}")

print("\nTrainable layers:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(name, param.numel())

print("Model device:", next(model.parameters()).device)
```

```python
LEARNING_RATE = 3e-4
WEIGHT_DECAY = 1e-4

# Experiment 2: Final residual stage + classification head
trainable_parameters = [
    p for idx in [4, 6]
    for p in model.blocks[idx].parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_parameters,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

criterion = nn.CrossEntropyLoss()

print("Optimizer: AdamW")
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Loss function: CrossEntropyLoss")

print("\n--- Tuning Targets (Experiment 2) ---")
print(
    "Block 4 trainable:",
    any(p.requires_grad for p in model.blocks[4].parameters())
)
print(
    "Block 6 trainable:",
    any(p.requires_grad for p in model.blocks[6].parameters())
)
print("Parameters in optimizer:", sum(
    p.numel() for p in trainable_parameters
))
```

```python

import time
import csv
import torch
import matplotlib.pyplot as plt
from pathlib import Path

MAX_EPOCHS = 20
PATIENCE = 6

# Save outputs locally in your MLP project folder
OUTPUT_DIR = Path.cwd() / "SlowFast_Experiment2"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

BEST_MODEL_PATH = OUTPUT_DIR / "best_model.pt"
HISTORY_PATH = OUTPUT_DIR / "training_history.csv"
LOSS_PLOT_PATH = OUTPUT_DIR / "training_vs_validation_loss.png"

# Mixed precision for CUDA
use_amp = device.type == "cuda"

scaler = torch.amp.GradScaler(
    "cuda",
    enabled=use_amp
)

print("Device:", device)
print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", PATIENCE)
print("Mixed precision:", use_amp)
print("Output folder:", OUTPUT_DIR.resolve())
```

```python
def create_slowfast_inputs(frames):
    # frames: [B, C, T, H, W]
    num_frames = frames.shape[2]

    indices = torch.linspace(
        0,
        num_frames - 1,
        steps=num_frames // 4,
        device=frames.device
    ).long()

    slow_pathway = torch.index_select(
        frames, 2, indices
    )

    fast_pathway = frames

    return [slow_pathway, fast_pathway]


# Verify the input shapes
sample_frames, sample_labels = next(iter(train_loader))
sample_frames = sample_frames.to(device)

slow, fast = create_slowfast_inputs(sample_frames)

print("Slow pathway:", slow.shape)
print("Fast pathway:", fast.shape)

del sample_frames, sample_labels, slow, fast
```

```python
def run_epoch(loader, training):
    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    # 1. Blanket rule: Set the entire network to evaluation mode first
    model.eval()

    # =====================================================================
    # CORRECTED FOR EXPERIMENT 2: Set target blocks to training mode
    # Dynamically loops through Outer Block 4 and Outer Block 6
    # =====================================================================
    if training:
        for idx in [4,6]:
            model.blocks[idx].train()
    # =====================================================================

    for frames, labels in loader:
        # Move tensors to GPU asynchronously alongside your pin_memory configuration
        frames = frames.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        # Structure input data into the custom [slow, fast] list structure
        inputs = create_slowfast_inputs(frames)

        if training:
            optimizer.zero_grad(set_to_none=True)

        # Control backpropagation tracking engine based on execution phase
        with torch.set_grad_enabled(training):
            # Implement Automatic Mixed Precision (AMP) to optimize VRAM footprint
            with torch.autocast(
                device_type=device.type,
                dtype=torch.float16,
                enabled=use_amp
            ):
                outputs = model(inputs)
                loss = criterion(outputs, labels)

            if training:
                # Scaler handles 16-bit decimal underflow protection
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

        # Aggregate training metrics accurately even if final batch is smaller
        batch_count = labels.size(0)
        total_loss += loss.item() * batch_count
        total_correct += (
            outputs.argmax(dim=1) == labels
        ).sum().item()
        total_samples += batch_count

    return (
        total_loss / total_samples,
        total_correct / total_samples
    )
```

```python
history = {
    "epoch": [],
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": []
}

best_val_loss = float("inf")
best_epoch = 0
epochs_without_improvement = 0

start_time = time.time()

for epoch in range(1, MAX_EPOCHS + 1):
    epoch_start = time.time()

    # Training pass
    train_loss, train_acc = run_epoch(
        train_loader,
        training=True
    )

    # Validation pass
    val_loss, val_acc = run_epoch(
        val_loader,
        training=False
    )

    history["epoch"].append(epoch)
    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_accuracy"].append(train_acc)
    history["val_accuracy"].append(val_acc)

    # Save history after every epoch
    with open(HISTORY_PATH, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(history.keys())

        for row in zip(*history.values()):
            writer.writerow(row)

    # Save the best model
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_epoch = epoch
        epochs_without_improvement = 0

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "best_val_loss": best_val_loss,
                "class_to_idx": CLASS_TO_IDX
            },
            BEST_MODEL_PATH
        )
    else:
        epochs_without_improvement += 1

    elapsed = (time.time() - epoch_start) / 60

    print(
        f"Epoch {epoch:02d}/{MAX_EPOCHS} | "
        f"Train loss: {train_loss:.4f} | "
        f"Val loss: {val_loss:.4f} | "
        f"Train acc: {train_acc:.4f} | "
        f"Val acc: {val_acc:.4f} | "
        f"Time: {elapsed:.2f} min"
    )

    if epochs_without_improvement >= PATIENCE:
        print(
            f"\nEarly stopping at epoch {epoch}."
        )
        print(f"Best epoch: {best_epoch}")
        break

print("\nTraining completed.")
print("Best epoch:", best_epoch)
print(f"Best validation loss: {best_val_loss:.4f}")
print(f"Total time: {(time.time() - start_time) / 60:.2f} minutes")
print("Best checkpoint:", BEST_MODEL_PATH)
```

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# --------------------------------------------------
# 1. Load the best trained checkpoint
# --------------------------------------------------
checkpoint = torch.load(
    BEST_MODEL_PATH,
    map_location=device,
    weights_only=False
)

model.load_state_dict(checkpoint["model_state_dict"])
model = model.to(device)
model.eval()

print("Checkpoint epoch:", checkpoint["epoch"])
print(f"Best validation loss: {checkpoint['best_val_loss']:.4f}")

# --------------------------------------------------
# 2. Test the model
# --------------------------------------------------
criterion = nn.CrossEntropyLoss()

all_labels = []
all_predictions = []
total_loss = 0.0
total_samples = 0

with torch.no_grad():
    for frames, labels in test_loader:
        frames = frames.to(device)
        labels = labels.to(device)

        # Use the same pathway function as training
        slow, fast = create_slowfast_inputs(frames)

        outputs = model([slow, fast])
        loss = criterion(outputs, labels)

        batch_size = labels.size(0)
        total_loss += loss.item() * batch_size
        total_samples += batch_size

        predictions = outputs.argmax(dim=1)

        all_labels.extend(labels.cpu().numpy())
        all_predictions.extend(predictions.cpu().numpy())

# --------------------------------------------------
# 3. Calculate test metrics
# --------------------------------------------------
if total_samples == 0:
    raise ValueError("Test dataset contains no samples.")
test_loss = total_loss / total_samples


accuracy = accuracy_score(
    all_labels, all_predictions
)

precision = precision_score(
    all_labels, all_predictions,
    average="macro", zero_division=0
)

recall = recall_score(
    all_labels, all_predictions,
    average="macro", zero_division=0
)

f1 = f1_score(
    all_labels, all_predictions,
    average="macro", zero_division=0
)

print("\n========== TEST RESULTS ==========")
print(f"Test samples:       {total_samples}")
print(f"Test loss:          {test_loss:.4f}")
print(f"Test accuracy:      {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Macro precision:    {precision:.4f}")
print(f"Macro recall:       {recall:.4f}")
print(f"Macro F1-score:     {f1:.4f}")

# --------------------------------------------------
# 4. Per-class performance
# --------------------------------------------------
idx_to_class = {
    index: name
    for name, index in CLASS_TO_IDX.items()
}

class_names = [
    idx_to_class[i]
    for i in range(len(idx_to_class))
]

print("\n========== CLASSIFICATION REPORT ==========")
print(classification_report(
    all_labels,
    all_predictions,
    labels=list(range(len(class_names))),
    target_names=class_names,
    digits=4,
    zero_division=0
))

# --------------------------------------------------
# 5. Confusion matrix
# --------------------------------------------------
cm = confusion_matrix(
    all_labels,
    all_predictions,
    labels=list(range(len(class_names)))
)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted class")
plt.ylabel("Actual class")
plt.title("SlowFast-R50 — Test Confusion Matrix (Experiment 2)")
plt.tight_layout()
plt.show()
```
